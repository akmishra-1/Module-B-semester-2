# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

### Subplots Advantage:
#### Using separate subplots makes it easier to isolate trends in each week independently. For example, if the patterns in Week 1 and Week 2 are different (e.g. a midweek peak vs a weekend peak), presenting them in separate panels avoids visual overlap and helps prevent confusion about which week a data point belongs to. This approach is ideal when you want to emphasize differences or changes week by week without clutter.

### Subplots Disadvantage:
#### However, subplots make direct comparison harder. If the y-axes aren't perfectly aligned or if scaling isn’t identical (which can easily happen), visual interpretation can be misleading. Even with identical axes, the eye has to bounce between two plots, making it more cognitively demanding to compare values day-by-day across weeks.

### Overlayed Plot Advantage
#### Overlaying both time series on the same plot makes direct, point by point comparison. It's easy to see, for example, that Wednesday had more visitors in Week 1 than Week 2, or that weekends were busier in one week. This approach is effective when you're trying to show trends over time or highlight changes across similar time intervals.

### Overlayed Plot Disadvantage
#### On the other side, overlaying plots can lead to clutter, especially if many time series are added or if their lines intersect frequently. Without clear labeling, color coding, or line styles the viewer could struggle to distinguish the series. It’s also harder to tell if there are subtle patterns are meaningful or just noise unless you're careful with formatting.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
# Figure 8.1, Page 188 

# data based on the chart
products = ['Product A', 'Product B', 'Product C', 'Product D', 'Product E']
years = ['2008', '2009', '2010', '2011', '2012', '2013', '2014']
values = {
    '2008': [390, 360, 0, 0, 0],
    '2009': [420, 400, 0, 0, 0],
    '2010': [430, 410, 100, 0, 0],
    '2011': [390, 370, 170, 150, 0],
    '2012': [300, 290, 190, 260, 0],
    '2013': [270, 260, 240, 220, 100],
    '2014': [250, 250, 180, 210, 210]
}

x = np.arange(len(products))
width = 0.11 # bar width

fig, ax = plt.subplots(figsize=(10, 6))

# plot each year's data
for i, year in enumerate(years):
    ax.bar(x + i*width - width*3, values[year], width, label=year)

# formatting
ax.set_title('Average Retail Product Price Per Year (Fig. 8.1 Pg. 188)', fontsize=14, weight='bold')
ax.set_ylabel('Price')
ax.set_xticks(x)
ax.set_xticklabels(products)
ax.set_ylim(0, 500)
ax.legend(title='Year')
ax.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
