# 2. Weekly graph question

Below are the graphs from last week's weekly graph question.  Suppose you wanted to cause the viewer to examine the last three months of data first.  How could you use preattentive attributes to do that?  Suggest several possibilities.

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series = 100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

## Answer

### If I wanted to make the last three data points stand out, we could use a bold contrasting color for those data points (red or bright blue), while dimming the rest (gray or muted tones). We can also shade the graph's background containing the last three months to emphasize the difference. 

### We can alos enlarge the markers (or change the shape e.g. circle to stars) or the line width for the last three data points because the larger graphical elements pull attention immediately.

### If we are using a preattentive attribute like the position on the page, we can place the modified grapphs at the top left corner of the page to grab the reader's attention immediately since people read left to right. Additionally, we can add in a label, note, or small arrows to call out the data at the last three months

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 99-119 of the Storytelling With Data book as best you can. (The first half of chapter four). You do not have to get the exact data values right, just the overall look and feel.

```python
# data from the chart
concerns = [
    "Engine power is less than expected",
    "Tires make excessive noise while driving",
    "Engine makes abnormal/excessive noise",
    "Seat material concerns",
    "Excessive wind noise",
    "Hesitation or delay when shifting",
    "Bluetooth system has poor sound quality",
    "Steering system/wheel has too much play",
    "Bluetooth system is difficult to use",
    "Front seat audio/entertainment/navigation controls"
]

values = [12.9, 12.3, 11.6, 11.6, 11.0, 10.3, 10.0, 8.8, 8.6, 8.2]

# same colors as the graph for each bar
color_map = {
    "Tires make excessive noise while driving": "#a74543",
    "Engine makes abnormal/excessive noise": "#a74543",
    "Excessive wind noise": "#a74543",
    "Steering system/wheel has too much play": "#b0b0ac",
    "Bluetooth system is difficult to use": "#b0b0ac",
    "Front seat audio/entertainment/navigation controls": "#b0b0ac"
}

# everything else is #daaca6 because i dont wanna type it out
colors = [color_map.get(c, "#daaca6") for c in concerns[::-1]]


# plot the bars with their colors, get the axes
fig, ax = plt.subplots(figsize=(8, 6))
bars = ax.barh(concerns[::-1], values[::-1], color=colors, edgecolor='none')

# white value labels inside the end of each bar, 0.3 spacing is best i think
for bar in bars:
    width = bar.get_width()
    ax.text(width - 0.3, bar.get_y() + bar.get_height() / 2,
            f'{width:.1f}', va='center', ha='right', fontsize=10, color='white')

# axis labels and stuff
ax.set_title("Top 10 design concerns (Storytelling with data pg 111, fig 4.9)", fontsize=14)
ax.set_xlabel("concerns per 1,000", fontsize=12)
ax.set_xlim(0, 14)

# removing the spines to make it clean
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_visible(False)
# remove the ticks
ax.tick_params(left=False, bottom=False, labelbottom=False)

plt.tight_layout()
plt.show()

```
